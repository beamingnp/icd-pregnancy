```python
from flask import Flask, render_template, request, send_file
import json
import re

app = Flask(__name__)

# بارگذاری دیتابیس بیماری‌ها
with open("diseases.json", "r", encoding="utf-8") as f:
    diseases = json.load(f)

# تابع ساده برای پیدا کردن بهترین کد بر اساس کلمات کلیدی مشترک
def guess_code(user_text):
    user_words = set(re.findall(r'\w+', user_text.lower()))
    if not user_words:
        return None

    best_match = None
    max_matches = 0

    for item in diseases:
        desc_words = set(re.findall(r'\w+', item['description'].lower()))
        # پیدا کردن کلمات مشترک بین شرح حال کاربر و توضیحات بیماری
        common_words = user_words.intersection(desc_words)

        if len(common_words) > max_matches:
            max_matches = len(common_words)
            best_match = item

    # اگر حداقل یک کلمه مشترک پیدا شد، نتیجه را برگردان
    return best_match if max_matches > 0 else None

@app.route('/')
def home():
    return render_template('index.html')

@app.route('/about')
def about():
    return render_template('about.html')

@app.route('/search', methods=['GET', 'POST'])
def search():
    result = None
    if request.method == 'POST':
        symptoms = request.form.get('symptoms', '')
        result = guess_code(symptoms)
        if result is None:
            result = False # یعنی جستجو انجام شد ولی چیزی پیدا نشد
    return render_template('search.html', result=result)

# دانلود فایل الفبایی (فایل JSON دیتابیس را به عنوان نمونه دانلود می‌کند)
@app.route('/download')
def download():
    return send_file('diseases.json', as_attachment=True)

# اجرای وب‌سایت روی پورت 5000 در پس‌زمینه
import threading
threading.Thread(target=app.run, kwargs={'port': 5000}).start()
```
